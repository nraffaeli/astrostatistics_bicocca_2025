**Monty Hall Problem** 

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 1000

victory = np.array([0, 1, 0])
victory_after_reveling_one = np.array([ 1, 0])
```

```python
switcher = np.zeros(N, dtype=int)
conservative = np.zeros(N, dtype=int)
newcomer = np.zeros(N, dtype=int)
```

```python
for i in range(N):
    random = np.random.choice(victory)
    if random == 0:
        switcher[i] = 1
        conservative[i] = 0
        newcomer[i] = np.random.choice(victory_after_reveling_one)
    elif random == 1:
        switcher[i] = 0
        conservative[i] = 1
        newcomer[i] = np.random.choice(victory_after_reveling_one)
    else:
        raise ValueError("Error: invalid random value")
    
switcher_counts = np.bincount(switcher, minlength=2)
conservative_counts = np.bincount(conservative, minlength=2)
newcomer_counts = np.bincount(newcomer, minlength=2)

bar_width = 0.25
x = np.arange(len(switcher_counts))

plt.bar(x - bar_width, switcher_counts, width=bar_width, label='Switcher', color='blue', edgecolor='black')
plt.bar(x, conservative_counts, width=bar_width, label='Conservative', color='green', edgecolor='black')
plt.bar(x + bar_width, newcomer_counts, width=bar_width, label='Newcomer', color='red', edgecolor='black')

plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram of Random Drawings')
plt.xticks(x, ['0', '1'])
plt.legend(loc='upper right')
plt.show()
```
