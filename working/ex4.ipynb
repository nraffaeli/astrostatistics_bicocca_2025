# Cloning Quasars 

Numerical simulations of the measurment process are often the only practical way to assess the complicated influences of selection effects, biases, and other processing stages of data taking. Notice that even without analytical PDFs and CDFs, we were able to accomplish the task.

```python
from astroML.datasets import fetch_dr7_quasar

data = fetch_dr7_quasar()

data = data[:10000]

z = data['redshift']
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
counts, edges =np.histogram(z,bins=np.linspace(0,10,101))

plt.figure(figsize=(10,5))
plt.hist(z, bins=edges, histtype='step', lw=2)
plt.xlabel('Redshift')
plt.ylabel("Counts of Quasars")
```

```python
plt.figure(figsize=(10,5))
plt.step(edges[1:],np.cumsum(counts),where='mid')
plt.xlabel('Redshift')
plt.ylabel('Cumulative Count')
plt.show()
```

```python
nsample = 10000
```

### Rejection Sampling



```python
centers = (edges[:-1] + edges[1:])/2
pdf = lambda val: np.interp(val, centers, counts / counts.max(), left=0, right=0)

x_random, y_random = np.random.uniform(0,10,size=nsample), np.random.uniform(0,1,size=nsample)

x_accepted , y_accepted = [], []
x_rejected , y_rejected = [], []

for i in range(nsample):
    if y_random[i] <= pdf(x_random[i]):
        x_accepted.append(x_random[i])
        y_accepted.append(y_random[i])
    else:
        x_rejected.append(x_random[i])
        y_rejected.append(y_random[i])

plt.scatter(x_accepted, y_accepted, color='blue', s=1)
plt.scatter(x_rejected, y_rejected, color='red', s=1)

plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python

plt.figure(figsize=(10,5))
plt.hist(x_accepted, bins=edges, density=True, histtype='step', lw=2, label='Accept/Reject Sampling PDF')
plt.xlabel('Redshift')
plt.ylabel('Probability density')
plt.legend()

plt.figure(figsize=(10,5))
plt.step(edges[1:],np.cumsum(counts)/np.sum(counts),where='mid',label='Original CDF')
plt.step(edges[1:],np.cumsum(np.histogram(x_accepted, bins=edges)[0])/len(x_accepted),where='mid',label='Accept/Reject Sampling CDF')
plt.xlabel('Redshift')
plt.ylabel('Cumulative Count')
plt.legend()
```

### Inverse transform

I'm using monotone-positivity of the cdf to ensure the convergence of the method.

```python
cdf = np.cumsum(counts)/np.sum(counts)

random_samples = np.random.uniform(0,1,size=nsample)

centers = (edges[:-1] + edges[1:])/2

for i in range(len(random_samples)):
    for j in range(len(cdf)):
        if random_samples[i] <= cdf[j]:
            random_samples[i] = centers[j]
            break

plt.figure(figsize=(10,5))
plt.hist(random_samples, bins=np.linspace(0,10,101), histtype='step', lw=2, label='Inverse Transform Sampling PDF')
plt.xlabel('Redshift')
plt.ylabel("Counts of Quasars")
plt.legend()

plt.figure(figsize=(10,5))
plt.step(edges[1:],np.cumsum(counts)/np.sum(counts),where='mid',label='Original CDF')
plt.step(edges[1:],np.cumsum(np.histogram(random_samples, bins=edges)[0])/len(random_samples),where='mid',label='Inverse Transform Sampling CDF')
plt.xlabel('Redshift')
plt.ylabel('Cumulative Count')
plt.legend()
```

### Geometric prior and selection effects

The theoretical curve is the geometric prior: it only accounts for the available comoving volume, assuming uniform density and perfect selection. Selection effects and quasar evolution are not included, so the observed distribution should not match it.

```python
import astropy
astropy.cosmology.Planck18

z_grid = np.linspace(0, 10, 1001)
dV_dz = astropy.cosmology.Planck18.differential_comoving_volume(z_grid).value
pdf_theory = dV_dz / np.trapz(dV_dz, z_grid)
```

```python
plt.figure(figsize=(10,5))
plt.hist(x_accepted, bins=edges, density=True, histtype='step', lw=2, label='Accept/Reject Sampling PDF')
plt.hist(random_samples, bins=np.linspace(0,10,101), density=True, histtype='step', lw=2, label='Inverse Transform Sampling PDF')
plt.plot(z_grid, pdf_theory, color='black', lw=2, label=r'Geometric Prior $p(z) \propto dV/dz$')
plt.xlabel('Redshift')
plt.ylabel('Probability density')
plt.legend()
```
